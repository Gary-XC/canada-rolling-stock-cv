## Financial Market Stock Data



```python
import yfinance as yf
```

```python
tickers = ["CNQ", "SU", "CVE"]
```

```python
df = yf.download(
    tickers=tickers,
    period="2mo",
    interval="1d",  
    group_by="ticker",  
    auto_adjust=True,  
    threads=True,  
)

assert df is not None
df.head()
```
